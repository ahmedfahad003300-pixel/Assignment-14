```python
#Q2
female_sleep = df.loc[df["gender"] == "Female", "sleep_hours"]
female_sleep_average=female_sleep.mean().round()

male_sleep = df.loc[df["gender"] == "Male", "sleep_hours"]
male_sleep_average=male_sleep.mean().round()

if female_sleep_average > male_sleep_average:
    print("females sleep more than men")
else:
    print("men sleep more than women")
```

```python
import pandas as pd

df = pd.read_csv("student_lifestyle_dataset.csv")
# Pearson (linear) and Spearman (rank) correlation using only pandas
pearson_r = df["screen_time"].corr(df["sleep_hours"])
spearman_r = df["screen_time"].corr(df["sleep_hours"], method="spearman")

print(f"Pearson correlation: {pearson_r:.4f}")
print(f"Spearman correlation: {spearman_r:.4f}")
# Compare average sleep across screen-time brackets
df["screen_group"] = pd.cut(
    df["screen_time"],
    bins=[0, 2, 4, 6],
    labels=["Low (1-2h)", "Medium (2-4h)", "High (4-6h)"],
)
print(df.groupby("screen_group", observed=False)["sleep_hours"].agg(["count", "mean", "median"]))
```

```python
#Q4
print(df["daily_transport_cost"].describe())
ax = df["daily_transport_cost"].plot.hist(
    bins=10,
    edgecolor="white",
    color="#2b5c8f",
    figsize=(8, 5),
    grid=True,)
ax.set_title("Histogram of Daily Transport Cost (pandas)")
ax.set_xlabel("Daily Transport Cost")
ax.set_ylabel("Frequency (Number of Students)")
ax.get_figure().tight_layout()
```

```python
#Q5
female_internet_average=df.loc[df["gender"]=="Female","internet_usage"].mean().round()
male_internet_average=df.loc[df["gender"]=="Male","internet_usage"].mean().round()

if female_internet_average > male_internet_average:
    print("females use more internet than men")
else:
    print("males use more internet than females")
    
```
